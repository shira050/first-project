# טעינת ספריות

נשתמש ב-OpenCV לקריאת התמונה וב-Matplotlib להצגתה. בהמשך נראה את ההבדל בין סדר הערוצים BGR של OpenCV לבין RGB שמצפה לו Matplotlib.


```python
import matplotlib.pyplot as plt
import cv2
```

# קריאת תמונה ובדיקת ממדים

הדגל `1` קורא תמונה צבעונית. `shape` מחזיר גובה, רוחב ושלושה ערוצי צבע; הסדר בפיקסל הוא BGR.


```python
bgr_img = cv2.imread('רמזור.png', 1)
bgr_img.shape
```

# הצגת מערך BGR ישירות

Matplotlib מפרשת את ערוץ 0 כאדום ואת ערוץ 2 ככחול, בניגוד לסדר של OpenCV. לכן הצגה ישירה עלולה לגרום לצבעים להיראות הפוכים.


```python
plt.imshow(bgr_img)
```

# היפוך ערוצי הצבע לתצוגה

החיתוך `[:, :, ::-1]` משאיר את כל הפיקסלים אך הופך את סדר שלושת הערוצים מ-BGR ל-RGB. כך Matplotlib מציגה את הצבעים הצפויים.


```python
plt.imshow(bgr_img[:, :, ::-1])
```

# הצגת ערוץ צבע יחיד

בחירת `[:, :, 0]` מציגה רק את הערוץ הראשון — כחול ב-BGR. `cmap="gray"` מציגה את עוצמת הערוץ כבהירות; התאים הבאים עושים זאת לערוצים הירוק והאדום.


```python
plt.imshow(bgr_img[:, :, 0], cmap='gray')
plt.title('Blue chanel')
```

```python
plt.imshow(bgr_img[:, :, 1], cmap='gray')
plt.title('Green chanel')
```

```python
plt.imshow(bgr_img[:, :, 2], cmap='gray')
plt.title('Red chanel')
```

# המרה מפורשת ל-RGB

`cv2.cvtColor` ממירה את התמונה לפי קוד צבע מוגדר. `COLOR_BGR2RGB` הופך את סדר הערוצים, והמערך שהתקבל מתאים ישירות להצגה ב-Matplotlib.


```python
rgb_image = cv2.cvtColor(bgr_img, cv2.COLOR_BGR2RGB)
plt.imshow(rgb_image)
```
